```python
import pandas as pd

# Yield curve flat at 5%
y = 0.05

# 1. Cash Flows & Prices
# B1: 15 years, semi-annual (30 periods), 5% annual coupon -> 2.5% per period
cf_b1 = erk.bond_cash_flows(15, 1000, 0.05, 2)
price_b1 = erk.bond_price(15, 1000, y, 0.05, 2)

# B2: 5 years, quarterly (20 periods), 6% annual coupon -> 1.5% per period
cf_b2 = erk.bond_cash_flows(5, 1000, 0.06, 4)
price_b2 = erk.bond_price(5, 1000, y, 0.06, 4)

# B3: 10 year Zero-Coupon (10 periods annual or 10 years single cash flow)
cf_b3 = erk.bond_cash_flows(10, 1000, 0.0, 1)
price_b3 = erk.bond_price(10, 1000, y, 0.0, 1)

print(f"Price B1: ${price_b1:.2f}")
print(f"Price B2: ${price_b2:.2f}")
print(f"Price B3: ${price_b3:.2f}")

# 2. Macaulay Durations (expressed in years by dividing by coupons_per_year)
dur_b1 = erk.macaulay_duration(cf_b1, y / 2) / 2
dur_b2 = erk.macaulay_duration(cf_b2, y / 4) / 4
dur_b3 = erk.macaulay_duration(cf_b3, y / 1) / 1  # or 10 for zero coupon

print(f"Duration B1: {dur_b1:.2f} years")
print(f"Duration B2: {dur_b2:.2f} years")
print(f"Duration B3: {dur_b3:.2f} years")

# 3. Liabilities: $100k at yr 3, $200k at yr 5, $300k at yr 10
liabilities = pd.Series(
    [0, 0, 100000, 0, 200000, 0, 0, 0, 0, 300000], index=range(1, 11)
)
# Compute duration of liabilities (approximate or via pv of cash flows)
```

```python
# Liabilities duration calculation (using 3, 5, 10 year timing)
liabilities = pd.Series(index=[3, 5, 10], data=[100000, 200000, 300000])
dur_liabilities = erk.macaulay_duration(liabilities, 0.05)
print(f"Duration of Liabilities: {dur_liabilities:.4f} years")

# Question 8: Weight of B2 in B1 + B2 portfolio
w_b2_b1b2 = (dur_liabilities - 10.73) / (4.37 - 10.73)
print(f"Weight of B2 (B1+B2): {w_b2_b1b2:.4f}")

# Question 10: Weight of B2 in B2 + B3 portfolio
w_b2_b2b3 = (dur_liabilities - 10.00) / (4.37 - 10.00)
print(f"Weight of B2 (B2+B3): {w_b2_b2b3:.4f}")
```
