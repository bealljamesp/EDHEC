```python
%load_ext autoreload
%autoreload 2
%matplotlib inline
import edhec_risk_kit as erk

ind = erk.get_ind_returns()
er = erk.annualized_rets(ind["1996":"2000"], 12)
cov = ind["1996":"2000"].cov()
```

```python
erk.msr(0.15, er, cov)
```

```python
erk.plot_ef(20, er, cov)
```

```python
ax = erk.plot_ef(20, er, cov)
# ax.set_xlim(left=0.02, right=0.15)
rf = 0.10
w_msr = erk.msr(rf, er, cov)
r_msr = erk.portfolio_return(w_msr, er)
vol_msr = erk.portfolio_vol(w_msr, cov)
# add the capital market line
cml_x = [0, vol_msr]
cml_y = [rf, r_msr]
ax.plot(
    cml_x, cml_y, color="green", marker="o", markersize=10, linewidth=2, label="CML"
)
```
